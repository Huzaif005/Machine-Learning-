```python
import pandas as pd

data = {
    "review": [
        "This movie is excellent",
        "I love this film",
        "The movie was amazing",
        "Very good acting",
        "I enjoyed this movie",
        "The story is wonderful",
        "This film is fantastic",
        "Great movie and great actors",
        "The movie is boring",
        "I hate this film",
        "Very bad acting",
        "The story is terrible",
        "I disliked this movie",
        "This film is disappointing",
        "The movie was awful",
        "Bad movie and bad actors",
        "The film is not good",
        "I regret watching this movie",
        "The movie is brilliant",
        "This is a terrible film"
    ],
    "sentiment": [
        "positive", "positive", "positive", "positive", "positive",
        "positive", "positive", "positive", "negative", "negative",
        "negative", "negative", "negative", "negative", "negative",
        "negative", "negative", "negative", "positive", "negative"
    ]
}

df = pd.DataFrame(data)
df.head()
```

```python
print("Total reviews:", len(df))
print(df["sentiment"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["review"],
    df["sentiment"],
    test_size=0.25,
    random_state=42,
    stratify=df["sentiment"]
)

print("Training reviews:", len(X_train))
print("Testing reviews:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Text converted into numerical features!")
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test_tfidf)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
review = input("Enter a movie review: ")

review_tfidf = tfidf.transform([review])
prediction = model.predict(review_tfidf)[0]

print("Predicted sentiment:", prediction.upper())
```
